```python
#Part B · Lab 2 — clean your own messy list

my_messylist=["nOuman ", " LAHoRe", "haHAHhahaHAha"]

for n in my_messylist:
    clean=n.strip().title().replace(" "," ")
    print(clean)
```

```python
#By trying each method individually, we can observe the following:

#- When we remove the .strip() method, the extra spaces at the beginning and end appear again.
#- When we remove the .title() method, the first letters of the words return to their original, inconsistent capitalization.

#What would happen if we put a comma inside one of the names, and how could we handle it?
#We will definately use .split function so it treats comma as a seperator


my_messylist=("nOuman ,  LAHoRe, haHAHhahaHAha")
solution=my_messylist.split(",")
print(solution)
```
